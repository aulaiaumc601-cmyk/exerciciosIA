## SciPy 3 - Raízes de x⁴ - 3x³ + 2

A função é escrita na forma necessária para que o método de raízes procure valores de x que façam o resultado chegar a zero.


```python
# Define a função do exercício.
def f(x):
    # Retorna x⁴ - 3x³ + 2.
    return x**4 - 3*x**3 + 2

# Cria vários chutes iniciais.
chutes = np.linspace(-3, 4, 30)

# Cria uma lista para guardar raízes diferentes.
raizes = []

# Percorre todos os chutes.
for chute in chutes:
    # Usa fsolve, do SciPy, para procurar uma raiz.
    raiz = fsolve(f, chute)[0]

    # Confere se a solução deixa a função próxima de zero.
    if np.abs(f(raiz)) < 0.000001:
        # Começa considerando que a raiz ainda não apareceu.
        repetida = False

        # Compara com as raízes já encontradas.
        for raiz_existente in raizes:
            # Verifica se a diferença é muito pequena.
            if np.abs(raiz - raiz_existente) < 0.00001:
                repetida = True

        # Adiciona somente uma vez cada raiz.
        if repetida == False:
            raizes.append(raiz)

# Exibe as raízes encontradas.
print("Raízes reais encontradas:", raizes)
```

## SciPy 16 - EDO de segunda ordem

A equação de segunda ordem precisa ser representada de uma forma que o solucionador de equações diferenciais consiga processar.


```python
# Importa o recurso necessário para esta etapa.
from scipy.integrate import solve_ivp

def sistema(x, Y):
    # Calcula ou armazena o valor de `y, dy`.
    y, dy = Y
    # Calcula ou armazena o valor de `ddy`.
    ddy = -y
    # Retorna o resultado calculado pela função.
    return [dy, ddy]

# Calcula ou armazena o valor de `x_eval`.
x_eval = np.linspace(0, 10, 100)
# Calcula ou armazena o valor de `solucao`.
solucao = solve_ivp(sistema, [0, 10], [1, 0], t_eval=x_eval)

# Exibe o resultado na saída do notebook.
print("x:", solucao.t)
# Exibe o resultado na saída do notebook.
print("y:", solucao.y[0])
```

```python
# Calcula ou armazena o valor de `n`.
n = 128
# Calcula ou armazena o valor de `t`.
t = np.arange(n) / n
# Calcula ou armazena o valor de `sinal`.
sinal = (
        np.sin(2*np.pi*5*t)
        + 0.5*np.cos(2*np.pi*10*t)
)
# Calcula ou armazena o valor de `transformada`.
transformada = fft(sinal)

# Exibe o resultado na saída do notebook.
print("Sinal:", sinal)
# Exibe o resultado na saída do notebook.
print("FFT:", transformada)
```

```python
# Importa o recurso necessário para esta etapa.


# Calcula ou armazena o valor de `x`.
x = np.random.random(100)
# Calcula ou armazena o valor de `y`.
y = np.random.random(100)
# Calcula ou armazena o valor de `z`.
z = np.random.random(100)

# Calcula ou armazena o valor de `fig`.
fig = plt.figure()
# Calcula ou armazena o valor de `ax`.
ax = fig.add_subplot(111, projection="3d")
# Configura ou exibe o gráfico.
ax.scatter(x, y, z)
# Configura ou exibe o gráfico.
ax.set_title("Dispersão 3D")
# Configura ou exibe o gráfico.
ax.set_xlabel("X")
# Configura ou exibe o gráfico.
ax.set_ylabel("Y")
# Configura ou exibe o gráfico.
ax.set_zlabel("Z")
# Exibe o resultado na saída do notebook.
print("Gráfico gerado com sucesso.")
# Configura ou exibe o gráfico.
plt.show()
```

## Matplotlib 16 - Anotação em x=5

Depois de traçar a linha, o ponto correspondente a x=5 recebe o texto pedido.


```python
# Calcula ou armazena o valor de `x`.
x = np.linspace(0, 10, 100)
# Calcula ou armazena o valor de `y`.
y = x**2

# Configura ou exibe o gráfico.
plt.plot(x, y)
# Calcula ou armazena o valor de `plt.title("Função y`.
plt.title("Função y = x²")
# Configura ou exibe o gráfico.
plt.xlabel("Eixo X")
# Configura ou exibe o gráfico.
plt.ylabel("Eixo Y")

# Calcula ou armazena o valor de `x_ponto`.
x_ponto = 5
# Calcula ou armazena o valor de `y_ponto`.
y_ponto = x_ponto**2
# Configura ou exibe o gráfico.
plt.annotate(
    # Calcula ou armazena o valor de `"x`.
    "x = 5",
    # Calcula ou armazena o valor de `xy`.
    xy=(x_ponto, y_ponto),
    # Calcula ou armazena o valor de `xytext`.
    xytext=(6, 20),
    # Calcula ou armazena o valor de `arrowprops`.
    arrowprops={"arrowstyle": "->"}
)

# Exibe o resultado na saída do notebook.
print("Gráfico gerado com sucesso.")
# Configura ou exibe o gráfico.
plt.show()
```

## Matplotlib 17 - Salvar gráfico em PDF

O gráfico é criado normalmente e, em vez de apenas ser exibido, também é salvo em formato PDF.


```python
# Calcula ou armazena o valor de `x`.
x = np.linspace(0, 10, 100)
# Calcula ou armazena o valor de `y`.
y = np.sin(x)

# Configura ou exibe o gráfico.
plt.plot(x, y)
# Configura ou exibe o gráfico.
plt.title("Gráfico salvo em PDF")
# Configura ou exibe o gráfico.
plt.xlabel("x")
# Configura ou exibe o gráfico.
plt.ylabel("sin(x)")
# Configura ou exibe o gráfico.
plt.savefig("grafico.pdf")
# Configura ou exibe o gráfico.
plt.close()

# Exibe o resultado na saída do notebook.
print("Arquivo grafico.pdf salvo.")
```

## Matplotlib 18 - Barras positivas e negativas

Os sinais dos valores são usados para separar visualmente números positivos e negativos por cor.


```python
# Calcula ou armazena o valor de `categorias`.
categorias = ["A", "B", "C", "D", "E"]
# Calcula ou armazena o valor de `valores`.
valores = np.array([10, -5, 7, -8, 12])
# Calcula ou armazena o valor de `cores`.
cores = ["green" if v >= 0 else "red" for v in valores]

# Calcula ou armazena o valor de `plt.bar(categorias, valores, color`.
plt.bar(categorias, valores, color=cores)
# Calcula ou armazena o valor de `plt.axhline(0, color`.
plt.axhline(0, color="black", linewidth=0.8)
# Configura ou exibe o gráfico.
plt.title("Valores Positivos e Negativos")
# Exibe o resultado na saída do notebook.
print("Gráfico gerado com sucesso.")
# Configura ou exibe o gráfico.
plt.show()
```

## Matplotlib 19 - Duas séries com legenda

As duas séries são desenhadas juntas e a legenda indica qual linha corresponde a cada uma.


```python
# Calcula ou armazena o valor de `x`.
x = np.linspace(0, 10, 100)
# Calcula ou armazena o valor de `y1`.
y1 = x
# Calcula ou armazena o valor de `y2`.
y2 = x**2

# Calcula ou armazena o valor de `plt.plot(x, y1, label`.
plt.plot(x, y1, label="y=x")
# Calcula ou armazena o valor de `plt.plot(x, y2, label`.
plt.plot(x, y2, label="y=x²")
# Configura ou exibe o gráfico.
plt.legend()
# Configura ou exibe o gráfico.
plt.title("Duas séries")
# Exibe o resultado na saída do notebook.
print("Gráfico gerado com sucesso.")
# Configura ou exibe o gráfico.
plt.show()
```

## Matplotlib 20 - Grade com espaçamento personalizado

A grade é ajustada para acompanhar um espaçamento definido entre as marcações do gráfico.


```python
# Cria os valores de x.
x = np.linspace(0, 10, 100)

# Calcula y = x².
y = x**2

# Exibe alguns valores numéricos.
print("Primeiros valores de x:", x[:5])
print("Primeiros valores de y:", y[:5])

# Cria o gráfico.
plt.plot(x, y)

# Define as marcações do eixo x com passo 1.
plt.xticks(np.arange(0, 11, 1))

# Define as marcações do eixo y com passo 10.
plt.yticks(np.arange(0, 101, 10))

# Adiciona uma grade.
plt.grid(True)

# Adiciona título e rótulos.
plt.title("y = x² com grade personalizada")
plt.xlabel("x")
plt.ylabel("y")

# Exibe o gráfico.
plt.show()
```
