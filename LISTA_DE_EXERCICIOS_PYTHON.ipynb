```python
#Implemente a busca binária para encontrar um elemento em uma lista ordenada
def busca(lista,valor):
  inicio=0
  fim= len(lista)-1

  while inicio <=fim:
      meio= (inicio+fim)//2

      if lista[meio]==valor:
        return meio

      elif valor>lista[meio]:
        inicio=meio+1
      else:
        fim=meio-1

  return -1

lista=[1,2,3,4,5,6]

print(busca(lista,4))


# O programa usa busca binária para encontrar um número dentro de uma lista ordenada.

# Passo a passo:

# A função busca() recebe a lista e o valor que será procurado. inicio e fim definem o começo e o final da lista. meio encontra o elemento do meio. Se o valor for encontrado, retorna a posição. Se o valor for maior, procura na metade da direita. Se for menor, procura na metade da esquerda. Se não encontrar, retorna -1. No exemplo, o número 4 está na posição 3, então o resultado é 3.
```

Optou-se pela busca binária por ser uma forma eficiente de procurar um valor em uma lista ordenada. A lista é dividida pela metade a cada tentativa, reduzindo a quantidade de elementos que precisam ser verificados.

```python
#implemente uma função para validar um CPF brasileiro
def validar_cpf(cpf):

    cpf = cpf.replace(".", "").replace("-", "")

    if len(cpf) != 11 or not cpf.isdigit():
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0

    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11

    if resto < 2:
        digito1 = 0
    else:
        digito1 = 11 - resto

    if digito1 != int(cpf[9]):
        return False

    soma = 0

    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11

    if resto < 2:
        digito2 = 0
    else:
        digito2 = 11 - resto

    if digito2 != int(cpf[10]):
        return False

    return True

cpf = input("Digite seu CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")





# O programa verifica se um CPF brasileiro é válido.

# Passo a passo:

# Remove os pontos e o - do CPF. Verifica se o CPF tem 11 números. Impede CPFs com todos os números iguais. Calcula o primeiro dígito do CPF. Confere se o primeiro dígito está correto. Calcula e confere o segundo dígito. Se estiver tudo certo, mostra "CPF válido". Caso contrário, mostra "CPF inválido".
```

Optou-se por criar uma função para concentrar todas as validações do CPF. As verificações foram feitas em etapas para garantir o formato correto e os dois dígitos verificadores, tornando a validação mais organizada e confiável.

```python
#implemente o algoritimo de ordenação por bolha(Bubble Sort).
lista = [4,2,3,1]

for i in range(len(lista)):
    for j in range(len(lista) - 1):
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]

print(lista)

# O programa usa o Bubble Sort para colocar os números da lista em ordem crescente.

# Passo a passo:

# A lista começa com [4, 2, 3, 1]. O for compara os números da lista. if lista[j] > lista[j + 1] verifica se o número da esquerda é maior. Se for, os dois números trocam de lugar. Isso se repete até a lista ficar ordenada. No final, o resultado é [1, 2, 3, 4].
```

```python
#Implemente uma função recursiva para calcular o fatorial de um número.
n=int(input("Digite um número: "))

def fatorial(n):
    # Caso base
    if n == 0 or n == 1:
        return 1
    # Passo recursivo
    else:
        return n * fatorial(n - 1)

print(fatorial(n))

#  programa calcula o fatorial de um número usando uma função recursiva.

# Passo a passo:

# O usuário digita um número. A função fatorial() recebe esse número. Se o número for 0 ou 1, retorna 1. Caso contrário, multiplica o número pelo fatorial do número anterior. A função continua chamando ela mesma até chegar em 1. print() mostra o resultado.
```

Optou-se por usar dois `for` para percorrer a lista e contar as ocorrências de cada número. Essa abordagem é simples e permite comparar cada valor com os demais sem usar funções prontas.


```python
#Implemente o algoritimo de ordenação por inserção(insertion Sort).
numeros = [5, 2, 4, 6, 1, 3]

for i in range(1, len(numeros)):

    chave = numeros[i]
    j = i - 1

    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = chave

print(numeros)

#O programa usa o Insertion Sort para colocar os números em ordem crescente.

#Passo a passo:
#A lista começa com vários números fora de ordem.
#chave guarda o número que será colocado no lugar certo.
#O while compara a chave com os números anteriores.
#Os números maiores são deslocados para a direita.
#A chave é colocada na posição correta.
#Isso se repete até toda a lista ficar ordenada. No final, o resultado é [1, 2, 3, 4, 5, 6].
```

Optou-se pelo Insertion Sort para ordenar a lista sem utilizar funções prontas. O `while` permite deslocar os valores maiores e inserir cada número na posição correta de forma simples.


```python
#Calcule a soma dos digitos de um número inteiro fornecido.
numero = int(input("Digite um número inteiro: "))

soma = 0

while numero > 0:
    digito = numero % 10
    soma += digito
    numero = numero // 10

print("Soma dos dígitos:", soma)

#O programa calcula a soma dos dígitos de um número inteiro.
#Passo a passo:

#O usuário digita um número.
#numero % 10 pega o último dígito.
#Esse dígito é adicionado à variável soma.
#numero // 10 remove o último dígito.
#O while continua até não sobrarem mais dígitos. No final, print() mostra a soma.
```

Optou-se por usar `while` para percorrer os dígitos do número um por um. O `% 10` e o `// 10` permitem pegar e remover o último dígito, tornando o cálculo simples e direto.


```python
#Cálculo de Frequência de Letras.
texto = input("Digite um texto: ")

letras = []
frequencias = []

for letra in texto:

    if letra not in letras:
        letras.append(letra)

        quantidade = 0

        for l in texto:
            if l == letra:
                quantidade += 1

        frequencias.append(quantidade)

for i in range(len(letras)):
    print(letras[i], "aparece", frequencias[i], "vezes")


# O programa conta quantas vezes cada letra aparece em um texto.

# Passo a passo:

# O usuário digita um texto. letras guarda as letras que já foram encontradas. frequencias guarda a quantidade de vezes que cada letra aparece. O primeiro for passa por cada letra do texto. Se a letra ainda não estiver na lista, o segundo for conta quantas vezes ela aparece. A quantidade é guardada em frequencias. No final, o programa mostra cada letra e sua frequência.
```

Optou-se por usar duas listas para armazenar as letras e suas respectivas frequências. Os dois `for` permitem identificar cada letra e contar suas ocorrências de forma simples, sem usar estruturas prontas.
